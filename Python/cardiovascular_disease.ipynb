```python
## het te mergen bestand inladen
df_popbu = pd.read_csv(
    r'C:\\Users\\asarb\\df_pop.csv',
    sep=';')
```

```python
## death rate from cardiovascular diseases and tot_ages zjn beide float. Ze kunnen berekeningen ondergaar
dfbu.info()
```

```python
##maar eerst de kolomnamen hernoemen
dfbu= dfbu.rename(columns = {'Tot_Ages':'Population'})
dfbu= dfbu.rename(columns = {'Entity':'Country'})
dfbu= dfbu.rename(columns = {'Age-standardized death rate from cardiovascular diseases among both sexes':'Death_rateCVD'})
dfbu.index.name = 'Index'
```

```python
#kolom aanmaken die het percentage sterfgevallen aan cardiovasculaire ziektes weergeven
##Het blijkt dat het geen nut heeft death rate te delen door het populatiegetal. Death is al per 100 000 mensen

## onderstaande kolom Perc_DeathRate_CVD vervalt dus
##dfbu['Perc_DeathRate_CVD'] = dfbu['Age-standardized death rate from cardiovascular diseases among both sexes']/ dfbu['Population']
```

```python
##Het bestand heeft nu wel missing values in kolom 'Perc_DeathRate_CVD'. Deze regels verwijderen we
dfbuWM = dfbu.dropna()
```

```python
#controleren of er nog missing values aanwezig zijn. Deze zijn er dus niet

dfbuWM.isna().any().any()
print(dfbuWM.isna().sum())
```

```python
dfbuWM['Year'].unique()
```

```python
#top 50 landen met de hoogste gemiddelde death rate vanaf 2020 (laatste jaar waar data voor beschikbaar was)
(dfbuWM[
    dfbuWM['Year']>= 2020
    ]
    .groupby('Country')['Death_rateCVD']
    .mean()
    .sort_values(ascending = False)
    .head(50))
```

```python
df_CardioVD = dfbuWM.copy()
```

```python
df_CardioVD.info()
```

```python
##Alvorens bestand op te slaan als .csv converteren we de death rate en tot populatie per rij naar datatype integer

df_CardioVD['Death_rateCVD'] = df_CardioVD['Death_rateCVD'].astype('int64')
df_CardioVD['Population'] = df_CardioVD['Population'].astype('int64')
```
