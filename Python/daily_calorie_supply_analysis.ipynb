```python
##we maken een nwe kolom aan die aangeeft welke rij tot welke categorie behoort. 
## 1. afzonderlijke landen (code met drie karakters) kriigen de categorie 'afzonderlijke landen'
## 2. indeling colgens FAO (Nan code/FAO in countrynaam)  krijgen de categoreie FOA, 
##en 3 indeling volgens Our worldin data )OWID in code) OWID

dfbu.loc[dfbu['Length_Code'] == 8.0, 'Indeling'] = 'OWID'
dfbu.loc[dfbu['Length_Code'] == 3.0, 'Indeling'] = 'afzonderlijke landen'
dfbu.loc[dfbu['Length_Code'].isna(), 'Indeling'] = 'FAO'
```

```python
## extra kolom maken die aangeeft tot welke kwartiel de rij behoort

#dfbu.loc[dfbu['Daily calorie supply per person'] < 2186.700200, 'Calorie_supply_quartile'] = '1st_qrt'
#dfbu.loc[dfbu['Daily calorie supply per person'] (>= 2186.700200 & < 2587.849900), 'Calorie_supply_quartile'] = '2nd_qrt'
#dfbu.loc[dfbu['Daily calorie supply per person'] (>= 2587.849900 & < 3041.689500), 'Calorie_supply_quartile'] = '3rd_qrt'
#dfbu.loc[dfbu['Daily calorie supply per person']  >= 3041.68950, 'Calorie_supply_quartile'] = '4th_qrt'

dfbu.loc[
    dfbu['Daily calorie supply per person'] < 2186.700200,
    'Calorie_supply_quartile'
] = '1st_qrt'

dfbu.loc[
    (dfbu['Daily calorie supply per person'] >= 2186.700200) &
    (dfbu['Daily calorie supply per person'] < 2587.849900),
    'Calorie_supply_quartile'
] = '2nd_qrt'

dfbu.loc[
    (dfbu['Daily calorie supply per person'] >= 2587.849900) &
    (dfbu['Daily calorie supply per person'] < 3041.689500),
    'Calorie_supply_quartile'
] = '3rd_qrt'

dfbu.loc[
    dfbu['Daily calorie supply per person'] >= 3041.689500,
    'Calorie_supply_quartile'
] = '4th_qrt'
```
